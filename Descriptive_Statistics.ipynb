Q1: Understanding Central Tendency
Data: \([10, 12, 11, 15, 14, 13, 12]\)

Mean: \(\frac{10+12+11+15+14+13+12}{7} = \frac{87}{7} \approx 12.43\)
Median: Sorted: \([10, 11, 12, 12, 13, 14, 15]\). Middle value = 12.
Mode: Most frequent value = 12.
Answer: The mean (~12.43) or median (12) are the most representative.
Since the data is closely packed without outliers, the mean effectively
represents the average daily sales

4  Median in Real Life
 Data: \([8,000, 9,500, 10,200, 11,000, 50,000]\)
 Median Calculation: Sorted middle value = $10,200.
 Why Median? The $50,000 car is an outlier. It would pull the "mean"
 up to $17,740, which doesn't reflect what most cars actually cost.The median ignore outliers.

 9 Combining Measures (Electricity Usage)
 Data: \([400, 420, 390, 450, 410]\)
 Mean: \(\frac{2070}{5} = \mathbf{414 \text{ kWh}}\).
 Standard Deviation (\(\sigma \)): \(\approx \mathbf{20.59}\text{\ kWh}\).
 Insight: The mean shows the typical monthly usage, while the low SD relative
 to the mean indicates the family has very consistent energy habits month-to-month.
